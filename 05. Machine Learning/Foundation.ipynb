
# **🍀 WHAT IS MACHINE LEARNING ?**

According to **Tom Mitchell's** "A computer program is said to learn from experience **E** with respect to some class of tasks **T** and performance measure **P**, if its performance at tasks in T, as measured by P, improves with experience E."

In plain words: a program gets better at a task the more experience (data) it gets.

- E (experience) = the data it practices on
- T (task) = the job, like spotting spam emails
- P (performance) = how well it does, like accuracy

So: more data → better results → that's learning.

**For Example:**



More Real life examples you see every day:

- 📧 **Spam filter** — sees millions of emails, learns which ones are junk, gets better as you mark spam.
- 📱 **Face unlock** — learns your face from many angles/photos to recognize you faster and more reliably.
- 🛒 **Amazon "you may also like"** — learns from millions of purchases to predict what you'll buy.
- 🚗 **Ride price estimates (Uber)** — learns from past trips, traffic, and demand to predict fare and arrival time.

Notice the pattern: none of these were *programmed with rules* like "if subject line says **WINNER**, mark spam" — they **learned the patterns from data**. That's the whole point of ML.

# 🍀 **AI** vs **ML** vs **DL**

**Artificial Intelligence** 

The field of computer science concerned with building systems that exhibit intelligence behaviour - i.e., the ability to perceive their environment , reason about it, and take actions that maximize the chance of achieving a goal.

***In simple terms*** :- Teaching a computer to be smart. Like: A robot dog that can fetch a ball because someone taught it how.

**Real Life Examples** 

- 🗣️ Alexa / Siri — you talk, it answers. Someone taught it to understand voices.
- 🤖 Robot vacuums — they find their way around your house without bumping into everything.

**Machine Learning**

The study of algorithms that improves their performance at a task by learning statistical patterns from data, without being explicitly programmed for that task.

***In simple terms*** :- Instead of telling the computer every rule, you show it lots of examples and it figures out the rules by itself. Like: You learn what a dog is by seeing lots of dogs - here nobody gave you a list of rules like "four legs + tail = dog."

**Real Life Examples**

- 📧 Spam filter — it saw millions of junk emails, so now it knows one when it shows up.
- 🛍️ Amazon "you may also like" — learned from tons of shoppers what people usually buy together.

**Deep Learning**

A subfield of machine learning that employs artificial neural networks with multiple hidden layers to automatically learn hierarchical feature representations from large volumes of data.

***In simple terms*** :- ML but with super big ***BRAIN*** (many layers. like a really deep stack of pancakes) that can handle super hard stuff. Like: Face unlock on your phone - it looks at your face and knows it's you, even in the dark.

**Real Life Examples**

- 🚗 Self-driving cars — they look at the road and recognize cars, people, traffic lights all by themselves.
- 💬 ChatGPT — read a gigantic amount of text, so it can chat with you about almost anything.

***IN Short***

- AI = "I want the computer to be smart."
- ML = "I'll show it examples so it learns."
- DL = "I'll use a giant brain-thing so it can learn even harder things."
